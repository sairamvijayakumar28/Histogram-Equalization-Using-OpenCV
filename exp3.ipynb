```python
# Import required libraries

import cv2
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
```

```python
img = cv2.imread('parrot.jpg', cv2.IMREAD_GRAYSCALE)
```

```python

img.shape
```

```python
# Display the grayscale image.

plt.imshow(img, cmap='gray')
plt.title('Original Image')
plt.show()

```

```python
# Plot the histogram of the grayscale image
# Display the images
plt.hist(img.ravel(),256,range = [0, 256]);
plt.title('Original Image')
plt.show()
```

```python
# Perform histogram equalization
# Equalize histogram
img_eq = cv2.equalizeHist(img)

```

```python


# Display the images.
plt.hist(img_eq.ravel(), 256, range = [0, 256])
plt.title('Equalized Histogram')

```

```python
# Display [1] the Original Image (Gray Image) and its Histogram, and [2] the Enhanced Image and its Histogram using a 2×2 layout in Matplotlib.



# Display the images.
plt.imshow(img_eq, cmap='gray')
plt.title('Original Image')
plt.show()


```

```python
img = cv2.imread('parrot.jpg', cv2.IMREAD_COLOR)
```

```python
 #Convert to HSV.
img_hsv = cv2.cvtColor(img, cv2.COLOR_BGR2HSV)
```

```python
# Perform histogram equalization
img_hsv[:,:,2] = cv2.equalizeHist(img_hsv[:, :, 2])
```

```python
# Convert back to BGR format
img_eq = cv2.cvtColor(img_hsv, cv2.COLOR_HSV2BGR)
```

```python
# Display
plt.imshow(img_eq[:,:,::-1]); plt.title('Equalized Image');plt.show()


```

```python
# Plot the histogram of colour image
plt.hist(img_eq.ravel(),256,range = [0, 256]); plt.title('Histogram Equalized');plt.show()

```
